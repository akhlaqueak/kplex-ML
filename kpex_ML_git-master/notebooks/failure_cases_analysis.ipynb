```python
import pandas as pd

# Load the Excel file and the specific sheet into a DataFrame
file_path = 'k-Plex_ML_Failure_Cases.xlsx'
sheet_name = '0% Tol'  # Make sure the sheet name matches exactly, including spaces and symbols

df = pd.read_excel(file_path, sheet_name=sheet_name)

# Display the first few rows of the DataFrame to verify
df.head()
```

```python
# Filter rows where:
#   with_holdout == True, strat_on == 'y', and k_value == 'K' (adjust if 'K' is numeric)
filtered_df = df[
    (df['with_holdout'] == True) & 
    (df['strat_on'] == 'y and K') # Replace 'K' with the actual value if different (e.g., 2, '2', etc.)
]

# Display the filtered DataFrame
filtered_df.head()
```

```python
filtered_df['source_model'].value_counts()
```

```python
import pandas as pd

def model_rows(model_name):
    model_df = filtered_df[filtered_df['source_model'] == model_name]
    return model_df

rows = model_rows('RF')
rows
```

```python
rows = model_rows('XGB')
rows
```

```python
def model_pair_rows(model1, model2):
    """
    Returns rows for model1 and model2, where model2 only includes rows that share
    the same (with_holdout, strat_on, k_value, dataset) combinations as model1.
    The result is sorted by these grouping columns.
    
    Args:
        model1 (str): First model name (e.g., 'RF', 'LSTM')
        model2 (str): Second model name
        
    Returns:
        pd.DataFrame: Combined and sorted rows for both models
    """
    # Get model1 rows
    model1_df = filtered_df[filtered_df['source_model'] == model1]
    
    # Get unique combinations of grouping columns from model1
    model1_groups = model1_df[['with_holdout', 'strat_on', 'k_value', 'dataset']].drop_duplicates()
    
    # Get model2 rows that match model1's group combinations
    model2_df = pd.merge(
        filtered_df[filtered_df['source_model'] == model2],
        model1_groups,
        on=['with_holdout', 'strat_on', 'k_value', 'dataset'],
        how='inner'
    )
    
    # Combine and sort the results
    combined_df = pd.concat([model1_df, model2_df]) \
                   .sort_values(by=['with_holdout', 'strat_on', 'k_value', 'dataset', 'source_model'])
    
    return combined_df

# Example usage:
rows = model_pair_rows('RF', 'SVC_Linear') #!!! this fixed RF's rec-epinions-user-ratings, as well as all LR models
rows
```

# Identifying RF/XGB “Surprise” Failures

Motivation:
RF and XGB are generally our strongest baselines, but we observed cases where all other models succeeded and only RF/XGB failed. To root‐cause these “surprise” failures, we needed a clean, end-to-end way to isolate exactly which datasets (and under which split/k-values) triggered them.

Aggregate all model outputs into two masters:

1. `combined_df_bad` includes our “weaker” models (SVC, LR variants, LSTM).

2. `combined_df_good` holds RF, XGB, and NN results.

3. Flag overlaps by joining on with_holdout, strat_on, k_value, dataset, annotating where a model’s failure appears in both sets.

4. Filter for rows where source_model is RF or XGB and is_in_bad == False, i.e. only RF/XGB missed but all others passed.

Summarize these by group into a final table with a failed_models list for each (inference_on, holdout, strat_on, k_value, dataset).

```python
import pandas as pd

def add_overlap_and_source(
    df_good: pd.DataFrame,
    df_bad:  pd.DataFrame,
    join_cols:       list = ['with_holdout','strat_on','k_value','dataset'],
    flag_col:        str  = 'is_in_bad',
    bad_source_col:  str  = 'source_model',
    new_source_col:  str  = 'source_model_from_bad',
):
    """
    Merge df_good with df_bad on join_cols, bringing in exactly one
    column new_source_col and a boolean flag.
    """
    # 1) prepare df_bad with the renamed source column
    bad_small = (
        df_bad[join_cols + [bad_source_col]]
        .drop_duplicates()
        .rename(columns={bad_source_col: new_source_col})
    )

    # 2) merge left onto df_good
    merged = df_good.merge(
        bad_small,
        on=join_cols,
        how='left'
    )

    # 3) flag where we did get a match
    merged[flag_col] = merged[new_source_col].notna()

    return merged
```

```python
def find_rf_xgb_failures(df):
    """
    From your combined DataFrame (with columns including
    source_model, source_model_from_bad, is_in_bad, etc.),
    return only the rows where:
      - source_model is RF or XGB
      - is_in_bad is False (i.e. they weren’t in the 'bad' set)
    """
    # 1) define which models you care about
    target_models = {"RF", "XGB", "NN"}

    # 2) filter
    mask = (
        df['source_model'].isin(target_models) &
        (~df['is_in_bad'].fillna(False))  # treat NaN as False
    )
    result = (
        df.loc[mask]
          .sort_values(['inference_on','with_holdout','strat_on','k_value','dataset'])
          .reset_index(drop=True)
    )
    return result

```

```python
def compile_failed_models(
    df: pd.DataFrame,
    group_cols = ['inference_on','with_holdout','strat_on','k_value','dataset'],
    keep_cols = [],
    source_col: str = 'source_model',
    new_col: str    = 'failed_models'
) -> pd.DataFrame:
    """
    From a fail-case table (one row per (group, source_model)),
    aggregate source_model into a list, and keep first value of additional cols.
    """
    # Only keep relevant columns
    cols_to_keep = group_cols + [source_col] + keep_cols
    sub = df[cols_to_keep].copy()

    # Build aggregation dict
    agg_dict = {source_col: lambda models: sorted(set(models))}
    for col in keep_cols:
        agg_dict[col] = 'first'  # or another reducer (e.g. 'mean', 'max')

    # Group and aggregate
    summary = (
        sub
        .groupby(group_cols, as_index=False)
        .agg(agg_dict)
        .rename(columns={source_col: new_col})
    )
    return summary
```

```python
# your fail‐cases df (multiple rows per dataset/model combo)
fail_df = combined_df_sorted.copy()

# Get one row per dataset‐slice, listing all the models that failed there:
fail_summary = compile_failed_models(fail_df)
```

```python
import pandas as pd

def filter_failures_by_models(
    fail_summary: pd.DataFrame,
    models_to_match: list = ["RF", "XGB", "NN"],
    group_cols:     list = ["inference_on", "with_holdout", "strat_on", "k_value", "dataset"],
    failed_col:     str  = "failed_models",
) -> pd.DataFrame:
    """
    Returns the subset of fail_summary where `failed_models` contains
    any of the names in `models_to_match`.  Keeps each row once,
    then sorts by group_cols.
    """
    # 1) Boolean mask: does the failed_models list intersect with models_to_match?
    mask = fail_summary[failed_col].apply(
        lambda lst: any(m in lst for m in models_to_match)
    )

    # 2) Filter and sort
    df_filtered = (
        fail_summary[mask]
        .drop_duplicates(subset=group_cols)  # ensure no overlaps
        .sort_values(by=group_cols)
        .reset_index(drop=True)
    )
    return df_filtered
```

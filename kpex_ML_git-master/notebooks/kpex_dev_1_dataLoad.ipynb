```python
timeout_replacement = 1800
file_path = r"..\input_data\results-filtered-dataset.txt"

import pandas as pd
import re 

def load_excel_data(file_path, timeout_replacement = 1800):    
    columns_to_replace = ['UMKP-S', 'UMKP-R', 'UMKP-P', 'UMKP+-S', 'UMKP+-R', 'UMKP+-P', 'kPEX', 'kPEX-S', 'kPEX-R', 'kPEX-P']
    dfs = []

    df = pd.read_csv(
      file_path,
      sep="\t",          # tab-separated
      quotechar='"',     # handles the quoted shell_sizes column
      engine="python"    # safer for complex text
   )

    df[columns_to_replace] = df[columns_to_replace].replace("×", timeout_replacement)
    # we will compute instance weights later by ourselves
    df.drop(columns=['Unnamed: 18',
       'Unnamed: 19', 'Unnamed: 20', 'Unnamed: 21', 'Unnamed: 22',
       'Unnamed: 23', 'Unnamed: 24', 'Unnamed: 25', 'Unnamed: 26',
       'Unnamed: 27', 'Unnamed: 28', 'Unnamed: 29'], inplace=True)
    df.rename(columns={"shell_sizes": "kcore_list", "core_number": "core_max", "k": "k_value"}, inplace=True)
    
    # Remove rows where ALL columns_to_replace equal timeout_replacement
    mask = (df[columns_to_replace] == timeout_replacement).all(axis=1)
    df = df[~mask]  # Keep rows where NOT all values are timeout_replacement
    
    dfs.append(df)

    # Concatenate all into one big DataFrame
    cat_df = pd.concat(dfs, ignore_index=True)
    dfs.reverse() # go from k = 2 to 20
    
    return cat_df, dfs # cat_df is the merged df, dfs is the array of df (df = DataFrame)
```

```python
cat_df, dfs = load_excel_data(file_path, timeout_replacement)
cat_df
```

```python
%%time

import pandas as pd

# 1. Load the local JSON file into a pandas DataFrame
business_df = pd.read_json("/shared/datasets/yelp/yelp_academic_dataset_business.json", lines=True)

# 2. Filter out rows where 'categories' is null and split/explode them
business_df = business_df[business_df['categories'].notna()]

# 3. Split categories into individual strings, trim whitespace, and explode into rows
categories = (
    business_df['categories']
    .str.split(',')           # Split into lists
    .explode()                # Explode into multiple rows
    .str.strip()              # Remove extra whitespace
    .drop_duplicates()        # Keep only unique categories
    .reset_index(drop=True)   # Reset index
)

# 4. Convert to DataFrame (single column 'category')
category_vertices = pd.DataFrame({'category': categories})

# 5. Save as CSV (without index column)
category_vertices.to_csv(
    "/shared/datasets/yelp/yelp_categories.csv",
    index=False,    # Skip writing row indices
    header=True     # Include column names
)

print(f"Saved {len(category_vertices)} unique categories to 'yelp_categories.csv'")
```

```python
%%time
import pandas as pd

# 1. Load the business data
business_df = pd.read_json("/shared/datasets/yelp/yelp_academic_dataset_business.json", lines=True)

# 2. Filter out rows with missing categories, remove duplicate category items
business_df = business_df[business_df['categories'].notna()]

def clean_categories(categories_str):
    if pd.isna(categories_str):
        return []
    categories_list = [c.strip() for c in categories_str.split(',')]
    return list(set(categories_list))

# 3. Split categories into lists, explode into rows, and trim whitespace
edges_df = (
    business_df[['business_id', 'categories']]
    .assign(categories=lambda x: x['categories'].apply(clean_categories))
    .explode('categories')                                       # One row per category
    .rename(columns={
        'business_id': 'business_id',
        'categories': 'category'
    })
)

# 4. Save as CSV (or JSON)
edges_df.to_csv(
    "/shared/datasets/yelp/category_edges.csv",
    index=False  # Skip row numbers
)

print(f"Saved {len(edges_df)} business-category edges to 'category_edges.csv'")
```

```python
%%time

import pandas as pd
import uuid

# 1. Load the JSON data (line-delimited format)
tip_df = pd.read_json("/shared/datasets/yelp/yelp_academic_dataset_tip.json", lines=True)

# 2. Add a unique UUID for each tip
tip_df["tip_id"] = [str(uuid.uuid4()) for _ in range(len(tip_df))]

# 3. Select and reorder columns
tip_df = tip_df[["tip_id", "user_id", "business_id", "text", "compliment_count", "date"]]

# 4. Save as CSV (without pandas index)
tip_df.to_csv(
    "/shared/datasets/yelp/yelp_tip.csv",
    index=False,      # Skip writing row numbers
    quoting=1,        # Quote non-numeric fields (e.g., text)
    escapechar='\\'   # Escape quotes in text if needed
)

print(f"✅ Saved {len(tip_df)} tips to CSV with unique IDs")
```

# Friend Edges

```python
!grep "6qGSOvsTgP1oIIU0kvLu3w" /shared/datasets/yelp/yelp_academic_dataset_user.json
```

We can see that --cxdcv_b9uhAKsKTlyC3Q is not in yelp_academic_dataset_user.json, so never appeared in src but can appear in dst. We should filter out such nodes.

```python
import json

src_set = set()

INPUT_PATH = "/shared/datasets/yelp/yelp_academic_dataset_user.json"

with open(INPUT_PATH, 'r') as f:
    for line in f:
        user = json.loads(line)
        user_id = user.get('user_id')
        src_set.add(user_id)
```

```python
len(src_set)
```

```python
import json

def count_friend_edges(file_path):
    total_edges = 0
    with open(file_path, 'r') as f:
        for line in f:
            user = json.loads(line)
            friends = user.get('friends', '').split(',')
            for friend in friends:
                friend = friend.strip() # this is important!!! some IDs have space before and after
                if friend in src_set:
                    total_edges += 1
    return total_edges

INPUT_PATH = "/shared/datasets/yelp/yelp_academic_dataset_user.json"
count_friend_edges(INPUT_PATH)
```

```python
%%time
import pandas as pd
import json
from tqdm import tqdm

# Configuration
INPUT_PATH = "/shared/datasets/yelp/yelp_academic_dataset_user.json"
OUTPUT_PATH = "/shared/datasets/yelp/friend_edges.csv"
CHUNKSIZE = 10000  # Rows per batch

# Initialize output file with headers
pd.DataFrame(columns=['src', 'dst']).to_csv(OUTPUT_PATH, index=False)

# Process in batches
with open(INPUT_PATH) as f:
    total_lines = sum(1 for _ in f)  # Get total lines for progress bar

with open(INPUT_PATH) as f, tqdm(total=total_lines, desc="Processing") as pbar:
    for chunk in pd.read_json(f, lines=True, chunksize=CHUNKSIZE):
        # Process each row individually
        rows = []
        for _, row in chunk.iterrows():
            if pd.notna(row['friends']):
                for friend in str(row['friends']).split(','):
                    friend = friend.strip()
                    if friend in src_set:
                        rows.append({
                            'src': row['user_id'],
                            'dst': friend.strip()
                        })
        
        # Write the batch
        if rows:
            pd.DataFrame(rows).to_csv(
                OUTPUT_PATH,
                mode='a',
                header=False,
                index=False,
                quoting=1
            )
        
        pbar.update(len(chunk))

print(f"\nProcessing complete! Saved to {OUTPUT_PATH}")

# runs for 7-8 min
```

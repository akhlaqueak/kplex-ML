```python
import pandas as pd
import ast
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
```

```python
# 1. Load data
excel_path = r"D:\Studies\Research\Da_Project\k-Plex ML\kplex_ML_data.xlsx"
df = pd.read_excel(excel_path, sheet_name="k=3", header=1)
df = df.drop(columns=["Unnamed: 0"])
df = df.drop(columns=["Unnamed: 8"])
```

```python
# Load all sheets into one Dataframe
import re 
xls = pd.ExcelFile(excel_path)
all_sheets = xls.sheet_names  # list of sheet names

# List to collect all DataFrames
dfs = []

# Iterate over each sheet
for sheet in all_sheets:
    df = pd.read_excel(excel_path, sheet_name=sheet, header=1)
    # df["k_value"] = sheet  # add new column with sheet name
    match = re.search(r"\d+", sheet)
    k_value = int(match.group()) if match else None
    df["k_value"] = k_value
    dfs.append(df)

# Concatenate all into one big DataFrame
final_df = pd.concat(dfs, ignore_index=True)

# Save to CSV
# final_df.to_csv("combined_data.csv", index=False)

# print("✅ Combined CSV saved as 'combined_data.csv'")
```

```python
def parse_core_list(cell):
    if isinstance(cell, str):
        try:
            return list(ast.literal_eval(cell))
        except (ValueError, SyntaxError):
            return []
    elif isinstance(cell, (list, tuple)):
        return list(cell)
    else:
        return []
```

```python
df.head()
```

```python
df["core_sizes"] = df["k_core, core_size list"].apply(parse_core_list)
df["core_min"] = df["core_sizes"].apply(lambda L: min(L) if len(L) > 0 else None)
df["core_max"] = df["core_sizes"].apply(lambda L: max(L) if len(L) > 0 else None)
```

```python

# def parse_list(cell):
#     if isinstance(cell, str):
#         return ast.literal_eval(cell)
#     return cell

# df["core_sizes"] = df["k_core, core_size list"].apply(parse_list)

# 3. Determine maximum number of tuples per row
max_cores = df["core_sizes"].apply(len).max()  # here, max_cores = 5

# 4. Expand into separate columns
for i in range(max_cores):
    # for each position i, create two new cols
    df[f"core_{i+1}_start"] = df["core_sizes"].apply(lambda L: L[i][0] if i < len(L) else 0)
    df[f"core_{i+1}_end"]   = df["core_sizes"].apply(lambda L: L[i][1] if i < len(L) else 0)
```

```python
# max_cores = df["cores"].apply(len).max()
df[df["core_sizes"].apply(len) ==150]["k_core, core_size list"]
```

```python
# For replacing x with 0 or 1800. Need to ask once again.
cols_to_clean = ['UMKP', 'AltRB', 'kPEX', 'Pivot_Br', 'S_Br']
df[cols_to_clean] = df[cols_to_clean].replace({"×": 0})
for col in cols_to_clean:
    df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0).astype(int)
```

```python
target_cols = ['UMKP', 'AltRB', 'kPEX', 'Pivot_Br', 'S_Br']
```

```python
all_cols = list(df.columns)
feature_cols = list(set(all_cols)-set(target_cols))
```

```python
# 1. Create a new column 'BestAlgo' that stores the column label with the minimum value per row
df['BestAlgo'] = df[target_cols].idxmin(axis=1)

# 2. (Optional) Inspect how many rows chose each algorithm
print(df['BestAlgo'].value_counts())
```

```python
cols_to_drop = target_cols + ['dataset', 'k_core, core_size list', 'core_sizes', 'core_min', 'core_max']
X = df.drop(columns=cols_to_drop + ['BestAlgo'])
y = df['BestAlgo']
```

```python
# 2. Identify columns whose dtype is 'object' or 'category'
obj_cols = X.select_dtypes(include=['object', 'category']).columns.tolist()
print("Non-numeric columns in X_train:", obj_cols)
```

```python
weights = df['(max-min)/min']
```

```python
X_train, X_test, y_train, y_test, w_train, w_test = train_test_split(
    X, y, weights, test_size=0.2, stratify=y, random_state=42
)
```

#### Random Forest

```python
pipeline_rf = Pipeline([
    ("scaler", StandardScaler()),  
    ("clf", RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1))
])

pipeline_rf.fit(X_train, y_train, clf__sample_weight=w_train)
```

```python
y_pred_rf = pipeline_rf.predict(X_test)
print("Random Forest Classification Report:")
print(classification_report(y_test, y_pred_rf))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf))
```

```python
param_grid = {
    "clf__n_estimators": [100, 200],
    "clf__max_depth": [None, 10, 20],
    "clf__min_samples_leaf": [1, 2, 5],
}

grid_rf = GridSearchCV(
    pipeline_rf, param_grid=param_grid, cv=5, n_jobs=-1, scoring="accuracy", verbose=1
)

grid_rf.fit(X_train, y_train, clf__sample_weight=w_train)  # :contentReference[oaicite:26]{index=26}
print("Best params (RF):", grid_rf.best_params_)
print("Best CV score (RF):", grid_rf.best_score_)
```

#### XGBoost

```python
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from xgboost import XGBClassifier

# 3. Label‐encode only on training labels
le = LabelEncoder()
y_train = le.fit_transform(y_train)  # maps strings to 0..K−1 :contentReference[oaicite:12]{index=12}

# ─── 4. For test labels, drop any unseen labels (unlikely if stratified)
mask_seen = y_test.isin(le.classes_)
X_test_final  = X_test[mask_seen]
y_test_final_str = y_test[mask_seen]
y_test = le.transform(y_test_final_str)  # integer‐encoded test labels

# (If you don’t want to drop unseen classes, you must handle them—e.g., assign to a special code.)

# ─── 5. Build a pipeline for scaling + XGBoost
pipeline_xgb = Pipeline([
    ("scaler", StandardScaler()),  # Trees don’t strictly need scaling, but often OK
    ("clf", XGBClassifier(
        n_estimators=200,
        max_depth=6,
        learning_rate=0.1,
        use_label_encoder=False,
        eval_metric="mlogloss",
        random_state=42
    ))
])

# ─── 6. Fit, passing in integer labels & sample weights
pipeline_xgb.fit(
    X_train, 
    y_train, 
    clf__sample_weight=w_train.loc[X_train.index]  # ensure alignment by index
)  # :contentReference[oaicite:13]{index=13}

# ─── 7. Predict on test set
y_pred = pipeline_xgb.predict(X_test_final)

# ─── 8. Decode integer predictions back to original strings
y_pred_str = le.inverse_transform(y_pred)

# ─── 9. Evaluation
print("Classification Report:")
print(classification_report(y_test_final_str, y_pred_str))
print("Confusion Matrix:")
print(confusion_matrix(y_test_final_str, y_pred_str))
```

```python
param_grid_xgb = {
    "clf__n_estimators": [100, 200],
    "clf__max_depth": [6, 10, 20],
    "clf__learning_rate": [0.01, 0.1],
    "clf__subsample": [0.8, 1.0],
    "clf__colsample_bytree": [0.8, 1.0]
}

# ─── 3. Set up GridSearchCV with 5-fold CV, accuracy scoring ──────────────────
grid_xgb = GridSearchCV(
    pipeline_xgb,
    param_grid=param_grid_xgb,
    cv=5,
    n_jobs=-1,
    scoring="accuracy",
    verbose=1
)

# ─── 4. Fit the grid search, passing in sample weights for XGBoost ─────────────
# (Assumes: X_train, y_train, and w_train are already defined; y_train must be integer-encoded.)
grid_xgb.fit(X_train, y_train, clf__sample_weight=w_train)

# ─── 5. Print best hyperparameters and CV score ────────────────────────────────
print("Best params (XGB):", grid_xgb.best_params_)
print("Best CV score (XGB):", grid_xgb.best_score_)
```

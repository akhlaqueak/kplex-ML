```python
timeout_replacement = 1800
excel_path = "kplex_ML_data.xlsx"

import pandas as pd
import re 

def load_excel_data(excel_path, timeout_replacement):
    xls = pd.ExcelFile(excel_path)
    all_sheets = xls.sheet_names  # list of sheet names
    columns_to_replace = ["UMKP", "AltRB", "kPEX", "Pivot_Br", "S_Br"]
    dfs = []

    # Iterate over each sheet
    for sheet in all_sheets:
        df = pd.read_excel(excel_path, sheet_name=sheet, header=1)
        df[columns_to_replace] = df[columns_to_replace].replace("×", timeout_replacement)
        # we will compute instance weights later by ourselves
        df.drop(columns=["Unnamed: 0", "Unnamed: 8", "max/min", "(max-min)/min", "min"], inplace=True)
        df.rename(columns={"Unnamed: 7": "kcore_list"}, inplace=True)
        df.drop(df.index[-1], inplace=True) # remove last row
        match = re.search(r"\d+", sheet)
        k_value = int(match.group())
        df["k_value"] = k_value # add new column with sheet name
        
        # Remove rows where ALL columns_to_replace equal timeout_replacement
        mask = (df[columns_to_replace] == timeout_replacement).all(axis=1)
        df = df[~mask]  # Keep rows where NOT all values are timeout_replacement
        
        dfs.append(df)

    # Concatenate all into one big DataFrame
    cat_df = pd.concat(dfs, ignore_index=True)
    dfs.reverse() # go from k = 2 to 20
    
    return cat_df, dfs # cat_df is the merged df, dfs is the array of df (df = DataFrame)
```

```python
cat_df, dfs = load_excel_data(excel_path, timeout_replacement)
cat_df
```

# Instance Weight Functions

```python
df = cat_df
y_columns = ["UMKP", "AltRB", "kPEX", "Pivot_Br", "S_Br"]
epsilon = 0.001 # need to set !!!
```

```python
# scheme1

def compute_weight(row):
    vals = row[y_columns]
    return vals.max() / vals.min() if vals.min() != 0 else vals.max() / epsilon

df["weight"] = df.apply(compute_weight, axis=1)
df
```

```python
# scheme2

def compute_weight(row):
    vals = row[y_columns]
    return (vals.max() - vals.min()) / vals.min() if vals.min() != 0 else vals.max() / epsilon

df["weight"] = df.apply(compute_weight, axis=1)
df
```

```python
# scheme3
import numpy as np

def compute_weight(row):
    vals = row[y_columns]
    return np.log(vals.max() / vals.min()) if vals.min() != 0 else np.log(vals.max() / epsilon)

df["weight"] = df.apply(compute_weight, axis=1)
df
```

```python
# scheme4
import numpy as np

def compute_weight(row):
    vals = row[y_columns]
    return np.log(1 + (vals.max() - vals.min()) / vals.min()) if vals.min() != 0 else np.log(1 + vals.max() / epsilon)

df["weight"] = df.apply(compute_weight, axis=1)
df
```

```python
# scheme5

def compute_weight(row):
    vals = row[y_columns]
    return vals.min()

df["weight"] = df.apply(compute_weight, axis=1)
df
```

```python
# scheme6
import numpy as np

def compute_weight(row):
    vals = row[y_columns]
    return np.log(1 + vals.min())

df["weight"] = df.apply(compute_weight, axis=1)
df
```

```python
# scheme7
import numpy as np

def compute_weight(row):
    vals = row[y_columns]
    speed_diff_score = np.log(vals.max() / vals.min()) if vals.min() != 0 else np.log(vals.max() / epsilon)
    return vals.min() * speed_diff_score

df["weight"] = df.apply(compute_weight, axis=1)
df
```

```python
# scheme8

import numpy as np

def compute_weight(row):
    vals = row[y_columns]
    speed_diff_score = np.log(1 + (vals.max() - vals.min()) / vals.min()) if vals.min() != 0 else np.log(1 + vals.max() / epsilon)
    return vals.min() * speed_diff_score

df["weight"] = df.apply(compute_weight, axis=1)
df
```

```python
# scheme9
import numpy as np

def compute_weight(row):
    vals = row[y_columns]
    speed_diff_score = np.log(vals.max() / vals.min()) if vals.min() != 0 else np.log(vals.max() / epsilon)
    log_min = np.log(1 + vals.min())
    return log_min * speed_diff_score

df["weight"] = df.apply(compute_weight, axis=1)
df
```

```python
# scheme10
import numpy as np

def compute_weight(row):
    vals = row[y_columns]
    speed_diff_score = np.log(vals.max() / vals.min()) if vals.min() != 0 else np.log(vals.max() / epsilon)
    log_min = np.log(1 + vals.min())
    return log_min * speed_diff_score

df["weight"] = df.apply(compute_weight, axis=1)
df
```

```python
df_sorted = df.sort_values(by="weight", ascending=False).reset_index(drop=True)
df_sorted
```

# Highlight wrong predictions from different models

In this script, we load the wrong prediction results from each results folder and highlight the rows where the time_gap_ratio, calculated as $(\text{predicted} - \text{fastest})/\text{fastest}$, is greater than or equal to 0.5 by marking them in yellow. These highlighted entries are then aggregated into a master table, which also includes the source file name as a column. This consolidated table will be used for subsequent comparisons against master dataframes from other models that were trained and tested under identical conditions.

Aggregated excel having rows with time gap ratio >= 0.5, which can be adjusted by parameter `threshold=0.5` 

```python
import write_utils
import pandas as pd
import numpy as np
```

## For Random Forest

```python
master_for_RF_expand_and_split = write_utils.highlight_and_collect_threshold(
    model_folder="RF_results/tie_tol/0/expand_and_split",
    filename_pattern="wrong_predictions_",
    threshold=0.5
)

master_for_RF_split_and_expand = write_utils.highlight_and_collect_threshold(
    model_folder="RF_results/tie_tol/0/split_and_expand",
    filename_pattern="wrong_predictions_",
    threshold=0.5
)
```

##### Aggregated excel having rows with time gap ratio >= 0.5

```python
# master_for_NN, master_for_RF, master_for_LR each with `filename` column
master_dict_RF = {
    'expand_and_split': write_utils.extract_filename_info(master_for_RF_expand_and_split),
    'split_and_expand': write_utils.extract_filename_info(master_for_RF_split_and_expand),}
```

```python
combined_df_RF = write_utils.aggregate_master_dataframes(
    master_dict_RF, 
    output_dir="./RF_results",
    save_files = False
)
combined_df_RF = combined_df_RF.rename(columns={'source_model': 'strategy'})
```

```python
cols = combined_df_RF.columns.tolist()
cols.remove("dataset")
cols.remove("strategy")
cols.remove("inference_on")
cols.remove("strat_on")
cols.remove("exp_num")

new_order = ["inference_on", "strat_on", "strategy", "dataset", "exp_num"] + cols

combined_df_RF = combined_df_RF[new_order]
# Sort by multiple columns in the specified order:
combined_df_sorted_RF = combined_df_RF.sort_values(
    by=["inference_on", "strat_on", "strategy", "dataset"],
    ascending=[True, True, True, True]  # change to False for any descending order
).reset_index(drop=True)

outfile = f'./RF_results/highlighted_errors_agg_best_experiments.xlsx'
combined_df_sorted_RF.to_excel(outfile, index=False)
```

# Compare highlighted rows from different models

General pipeline:

1. Accepts multiple master DataFrames like master_for_NN, master_for_RF, each with a filename column.

2. Combines them into aggregated DataFrames grouped by filename, i.e. for each unique filename, collects rows from all master DFs that have that filename.

3. Adds a new column indicating from which master DataFrame the row came (e.g. source_model = 'NN' / 'RF' / etc).

4. Writes these aggregated DFs into Excel files named like agg_<filename>_all_models.xlsx.

5. Each Excel file:

        - Sorted by dataset.

        - With columns ordered so that dataset is first, source_model is second, then the rest.

```python
master_dict = {
    'RF': combined_df_sorted_RF,
    'XGB': combined_df_sorted_XGB,
    'SVC_Linear': combined_df_sorted_SVC_linear,
    'SVC_RBF': combined_df_sorted_SVC_RBF,
    'LR_no_regularization': combined_df_sorted_LR_0,
    'LR_l1': combined_df_sorted_LR_L1,
    'LR_l2': combined_df_sorted_LR_L2,
    'LR_l1+l2': combined_df_sorted_LR_elasticnet,
    'NN_hard_labels': combined_df_sorted_NN,
    'NN_soft_labels': combined_df_sorted_NN_soft,
    'LSTM': combined_df_sorted_LSTM
}
```

```python
import pandas as pd
import os


def write_by_inference_on(df: pd.DataFrame, output_path: str):
    """
    Writes `df` out to an Excel file with one sheet per unique value of `inference_on`.
    """
    # Make sure the file’s parent directory exists
    os.makedirs(os.path.dirname(output_path) or '.', exist_ok=True)

    # Create the Excel writer
    with pd.ExcelWriter(output_path, engine='openpyxl') as writer:
        # For each unique 'inference_on' value, create a sheet
        for inf_on in df['inference_on'].unique():
            sheet_name = inf_on[:31]  # Excel sheet names max length = 31
            subset = df[df['inference_on'] == inf_on]
            subset.to_excel(writer, sheet_name=sheet_name, index=False)

    print(f"Wrote {len(df)} rows to '{output_path}' with sheets: {list(df['inference_on'].unique())}")

```

```python
combined_df_sorted = combined_df.sort_values(
    by=["inference_on",  "with_holdout", "strat_on", "k_value" , "dataset", "source_model"],
    ascending=[True, True, True, True, True, True]  # change to False for any descending order
).reset_index(drop=True)

outfile = f'./aggregated_outputs/highlighted_errors_agg_all_models_and_experiments_tol_000.xlsx'
# combined_df_sorted.to_excel(outfile, index=False)
# Usage:
write_by_inference_on(
    combined_df_sorted,
    output_path=outfile
)
```
